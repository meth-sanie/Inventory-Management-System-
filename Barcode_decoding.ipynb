```python
import sqlite3

conn = sqlite3.connect("products.db")

def barcode_decoder(b, conn):
    try:
        cur = conn.cursor()
        cur.execute("SELECT PRODUCT_NAME FROM products WHERE barcode = ?", (b,))
        result = cur.fetchone()
        if result:
            return result[0]
        else:
            print("Product not found. Add it:")
            product_name = input("Enter product name wi: ")
            product_size = input("Enter product size: ")
            cur.execute(
                "INSERT INTO products (PRODUCT_NAME, PRODUCT_SIZE, barcode) VALUES (?, ?, ?)",
                (product_name, product_size, b)
            )
            conn.commit()
            return "Product added"
    except sqlite3.Error as e:
        return f"Database error: {e}"


print(barcode_decoder("4792081048441", conn))
```
