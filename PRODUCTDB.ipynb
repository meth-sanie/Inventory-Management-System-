```python
import pandas as pd
import sqlite3

df = pd.read_excel("/products.xlsx", dtype={"barcode": str})
conn = sqlite3.connect("products.db")
df.to_sql("products", conn, if_exists="replace", index=False)
conn.execute("CREATE UNIQUE INDEX idx_barcode ON products(barcode)")
conn.commit()
conn.close()
```
